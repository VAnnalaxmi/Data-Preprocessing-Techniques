```python

# ============================================================
# DATA VALIDATION - ALL MAJOR TECHNIQUES
# Banking Customer Dataset
# ============================================================

import pandas as pd
import numpy as np
import re

# ------------------------------------------------------------
# 1. CREATE BANKING DATASET
# ------------------------------------------------------------

data = {
    'Customer_ID': [101, 102, 103, 104, 105, 106, 106, 108],
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'John', 'Meena', 'Meena', 'Arun'],
    'Age': [22, 35, np.nan, 45, 28, 150, 150, 40],
    'Salary': [25000, 45000, 30000, np.nan, 50000, 90000, 90000, -5000],
    'Gender': ['Female', 'Male', 'Female', 'Male', 'Male', 'Female', 'Female', 'Unknown'],
    'City': ['Hyderabad', 'Chennai', 'Hyderabad', 'Mumbai', 'Delhi', 'Chennai', 'Chennai', 'Unknown'],
    'Email': [
        'anu@gmail.com',
        'ravi@gmail.com',
        'priya@gmail.com',
        'kiran@gmail.com',
        'john@gmail.com',
        'meena@gmail.com',
        'meena@gmail.com',
        'arun@gmail'
    ],
    'Account_Type': ['Savings', 'Current', 'Savings', 'Savings',
                     'Current', 'Savings', 'Savings', 'Current']
}

df = pd.DataFrame(data)

print("ORIGINAL BANKING DATASET")
display(df)


# ------------------------------------------------------------
# 2. MISSING VALUE VALIDATION
# ------------------------------------------------------------

# Checks missing values in each column.

print("\n2. MISSING VALUE VALIDATION")

print(df.isnull().sum())

# Displays rows containing missing values.

missing_rows = df[df.isnull().any(axis=1)]

print("Rows with Missing Values:")
display(missing_rows)


# ------------------------------------------------------------
# 3. DUPLICATE VALIDATION
# ------------------------------------------------------------

# Identifies completely duplicated records.

print("\n3. DUPLICATE VALIDATION")

print("Duplicate Records:", df.duplicated().sum())

duplicate_rows = df[df.duplicated()]

display(duplicate_rows)

# Checks duplicate Customer_ID values.

print("Duplicate Customer IDs:",
      df['Customer_ID'].duplicated().sum())


# ------------------------------------------------------------
# 4. DATA TYPE VALIDATION
# ------------------------------------------------------------

# Checks whether columns have expected data types.

print("\n4. DATA TYPE VALIDATION")

print("Actual Data Types:")
print(df.dtypes)

# Checks whether Age is numeric.

print("Age is Numeric:",
      pd.api.types.is_numeric_dtype(df['Age']))

# Checks whether Salary is numeric.

print("Salary is Numeric:",
      pd.api.types.is_numeric_dtype(df['Salary']))

# Checks whether Customer_ID is numeric.

print("Customer ID is Numeric:",
      pd.api.types.is_numeric_dtype(df['Customer_ID']))


# ------------------------------------------------------------
# 5. RANGE VALIDATION
# ------------------------------------------------------------

# Checks whether Age falls within the allowed range.

print("\n5. RANGE VALIDATION")

invalid_age = df[
    (df['Age'] < 0) |
    (df['Age'] > 100)
]

print("Invalid Age Records:")
display(invalid_age)

# Checks whether Salary is negative.

invalid_salary = df[df['Salary'] < 0]

print("Invalid Salary Records:")
display(invalid_salary)


# ------------------------------------------------------------
# 6. FORMAT VALIDATION
# ------------------------------------------------------------

# Checks whether email addresses follow a basic email pattern.

print("\n6. FORMAT VALIDATION")

def valid_email(email):
    if pd.isna(email):
        return False

    pattern = r'^[\w.-]+@[\w.-]+\.[A-Za-z]{2,}$'

    return bool(re.match(pattern, str(email)))

df['Email_Valid'] = df['Email'].apply(valid_email)

print("Email Validation:")
display(df[['Name', 'Email', 'Email_Valid']])

# Finds invalid email addresses.

invalid_emails = df[~df['Email_Valid']]

print("Invalid Emails:")
display(invalid_emails)


# ------------------------------------------------------------
# 7. CATEGORICAL VALUE VALIDATION
# ------------------------------------------------------------

# Checks whether Gender contains only allowed values.

print("\n7. CATEGORICAL VALUE VALIDATION")

allowed_gender = ['Male', 'Female']

invalid_gender = df[
    ~df['Gender'].isin(allowed_gender)
]

print("Invalid Gender Values:")
display(invalid_gender)

# Checks whether City contains allowed values.

allowed_cities = ['Hyderabad', 'Chennai', 'Mumbai', 'Delhi']

invalid_city = df[
    ~df['City'].isin(allowed_cities)
]

print("Invalid City Values:")
display(invalid_city)

# Checks valid Account Types.

allowed_accounts = ['Savings', 'Current']

invalid_accounts = df[
    ~df['Account_Type'].isin(allowed_accounts)
]

print("Invalid Account Types:")
display(invalid_accounts)


# ------------------------------------------------------------
# 8. UNIQUENESS VALIDATION
# ------------------------------------------------------------

# Customer_ID should be unique for every customer.

print("\n8. UNIQUENESS VALIDATION")

print("Total Customer IDs:", len(df))

print("Unique Customer IDs:",
      df['Customer_ID'].nunique())

# Identifies records having repeated Customer_ID.

non_unique_ids = df[
    df['Customer_ID'].duplicated(keep=False)
]

print("Non-Unique Customer IDs:")
display(non_unique_ids)


# ------------------------------------------------------------
# 9. REFERENTIAL INTEGRITY VALIDATION
# ------------------------------------------------------------

# Creates a customer master table and a loan table.

customer_master = pd.DataFrame({
    'Customer_ID': [101, 102, 103, 104, 105]
})

loan_data = pd.DataFrame({
    'Loan_ID': [1, 2, 3, 4],
    'Customer_ID': [101, 102, 105, 110],
    'Loan_Amount': [50000, 100000, 75000, 60000]
})

# Checks whether every loan Customer_ID exists
# in the customer master table.

print("\n9. REFERENTIAL INTEGRITY VALIDATION")

invalid_foreign_keys = loan_data[
    ~loan_data['Customer_ID'].isin(
        customer_master['Customer_ID']
    )
]

print("Invalid Foreign Key Records:")
display(invalid_foreign_keys)


# ------------------------------------------------------------
# 10. CONSISTENCY VALIDATION
# ------------------------------------------------------------

# Checks whether related columns follow logical rules.
# Example: Age should be at least 18 for this bank's
# adult customer account policy.

print("\n10. CONSISTENCY VALIDATION")

inconsistent_age = df[
    (df['Age'] < 18) &
    (df['Age'].notna())
]

print("Customers Below Minimum Account Age:")
display(inconsistent_age)

# Checks whether account types use standard spelling.

print("Account Type Consistency:")
print(df['Account_Type'].value_counts())


# ------------------------------------------------------------
# 11. BUSINESS RULE VALIDATION
# ------------------------------------------------------------

# Checks business rules defined by the bank.

print("\n11. BUSINESS RULE VALIDATION")

# Rule 1: Customer must be at least 18 years old.

rule_age = df['Age'].between(18, 100)

print("Age Rule:")
print(rule_age)

# Rule 2: Salary must not be negative.

rule_salary = df['Salary'] >= 0

print("Salary Rule:")
print(rule_salary)

# Rule 3: Customer must have a valid account type.

rule_account = df['Account_Type'].isin(
    ['Savings', 'Current']
)

print("Account Type Rule:")
print(rule_account)

# Displays records violating at least one business rule.

business_rule_violations = df[
    ~(rule_age & rule_salary & rule_account)
]

print("Business Rule Violations:")
display(business_rule_violations)


# ------------------------------------------------------------
# 12. SCHEMA VALIDATION
# ------------------------------------------------------------

# Checks whether required columns exist.

print("\n12. SCHEMA VALIDATION")

required_columns = [
    'Customer_ID',
    'Name',
    'Age',
    'Salary',
    'Gender',
    'City',
    'Email',
    'Account_Type'
]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

extra_columns = [
    column for column in df.columns
    if column not in required_columns
]

print("Missing Required Columns:", missing_columns)
print("Extra Columns:", extra_columns)


# ------------------------------------------------------------
# 13. DATA COMPLETENESS VALIDATION
# ------------------------------------------------------------

# Calculates the percentage of available values in each column.

print("\n13. DATA COMPLETENESS VALIDATION")

completeness = (
    df[required_columns].notnull().mean() * 100
)

print("Column Completeness Percentage:")
display(completeness.to_frame(
    name='Completeness_Percentage'
))

# Calculates overall completeness.

overall_completeness = (
    df[required_columns].notnull().sum().sum()
    /
    (len(df) * len(required_columns))
) * 100

print("Overall Completeness:",
      round(overall_completeness, 2), "%")


# ------------------------------------------------------------
# 14. ACCURACY VALIDATION
# ------------------------------------------------------------

# Accuracy means checking values against a trusted source.
# Here, we compare customer details with a trusted reference.

print("\n14. ACCURACY VALIDATION")

trusted_data = pd.DataFrame({
    'Customer_ID': [101, 102, 103],
    'Expected_Name': ['Anu', 'Ravi', 'Priya']
})

accuracy_check = pd.merge(
    df,
    trusted_data,
    on='Customer_ID',
    how='inner'
)

accuracy_check['Name_Matches'] = (
    accuracy_check['Name'] ==
    accuracy_check['Expected_Name']
)

print("Accuracy Comparison:")
display(
    accuracy_check[
        ['Customer_ID', 'Name', 'Expected_Name', 'Name_Matches']
    ]
)


# ------------------------------------------------------------
# 15. FINAL DATA QUALITY REPORT
# ------------------------------------------------------------

print("\n====================================")
print("       FINAL DATA QUALITY REPORT")
print("====================================")

print("Total Records:", len(df))

print("Total Columns:", len(df.columns))

print("Missing Values:",
      df[required_columns].isnull().sum().sum())

print("Duplicate Records:",
      df.duplicated().sum())

print("Duplicate Customer IDs:",
      df['Customer_ID'].duplicated().sum())

print("Invalid Age Records:",
      ((df['Age'] < 0) | (df['Age'] > 100)).sum())

print("Negative Salary Records:",
      (df['Salary'] < 0).sum())

print("Invalid Emails:",
      (~df['Email_Valid']).sum())

print("Invalid Gender Records:",
      (~df['Gender'].isin(allowed_gender)).sum())

print("Invalid City Records:",
      (~df['City'].isin(allowed_cities)).sum())

print("Invalid Account Types:",
      (~df['Account_Type'].isin(allowed_accounts)).sum())

print("Overall Completeness:",
      round(overall_completeness, 2), "%")

print("\nDATA VALIDATION COMPLETED!")
```
