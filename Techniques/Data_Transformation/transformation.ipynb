```python

# ============================================================
# DATA TRANSFORMATION - ALL MAJOR TECHNIQUES
# Banking Customer Dataset
# ============================================================

import pandas as pd
import numpy as np

from sklearn.preprocessing import (
    MinMaxScaler,
    StandardScaler,
    RobustScaler,
    LabelEncoder,
    OneHotEncoder,
    OrdinalEncoder,
    PowerTransformer
)

# ------------------------------------------------------------
# 1. CREATE BANKING DATASET
# ------------------------------------------------------------

data = {
    'Customer_ID': range(101, 111),

    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'John',
             'Meena', 'Arun', 'Sita', 'Rahul', 'Divya'],

    'Age': [22, 35, 28, 45, 32, 40, 25, 50, 38, 29],

    'Salary': [25000, 45000, 30000, 65000, 40000,
               55000, 28000, 80000, 50000, 32000],

    'Balance': [10000, 25000, 15000, 50000, 20000,
                35000, 12000, 70000, 30000, 18000],

    'Loan_Amount': [0, 50000, 20000, 100000, 30000,
                    60000, 0, 150000, 40000, 10000],

    'Gender': ['Female', 'Male', 'Female', 'Male', 'Male',
               'Female', 'Male', 'Female', 'Male', 'Female'],

    'City': ['Hyderabad', 'Chennai', 'Delhi', 'Mumbai',
             'Hyderabad', 'Chennai', 'Delhi', 'Mumbai',
             'Hyderabad', 'Chennai'],

    'Education': ['Graduate', 'Postgraduate', 'Graduate',
                  'Postgraduate', 'Undergraduate', 'Graduate',
                  'Undergraduate', 'Postgraduate',
                  'Graduate', 'Undergraduate'],

    'Account_Type': ['Savings', 'Current', 'Savings',
                     'Current', 'Savings', 'Current',
                     'Savings', 'Current', 'Savings', 'Current'],

    'Joining_Date': [
        '2022-01-10', '2021-05-15', '2023-03-20',
        '2020-07-12', '2022-11-25', '2021-09-18',
        '2023-01-05', '2019-04-30', '2022-06-14',
        '2023-08-22'
    ]
}

df = pd.DataFrame(data)

print("ORIGINAL BANKING DATASET")
display(df)


# ------------------------------------------------------------
# 2. MIN-MAX NORMALIZATION
# ------------------------------------------------------------

# Converts numerical values into a range between 0 and 1.

minmax = MinMaxScaler()

df_minmax = df.copy()

df_minmax[['Age', 'Salary', 'Balance']] = (
    minmax.fit_transform(
        df_minmax[['Age', 'Salary', 'Balance']]
    )
)

print("\n2. MIN-MAX NORMALIZATION")
display(df_minmax[['Age', 'Salary', 'Balance']])


# ------------------------------------------------------------
# 3. Z-SCORE STANDARDIZATION
# ------------------------------------------------------------

# Converts values using mean and standard deviation.
# Resulting data has approximately mean 0 and standard deviation 1.

standard = StandardScaler()

df_standard = df.copy()

df_standard[['Age', 'Salary', 'Balance']] = (
    standard.fit_transform(
        df_standard[['Age', 'Salary', 'Balance']]
    )
)

print("\n3. Z-SCORE STANDARDIZATION")
display(df_standard[['Age', 'Salary', 'Balance']])


# ------------------------------------------------------------
# 4. DECIMAL SCALING
# ------------------------------------------------------------

# Divides values by a power of 10 to scale them.

df_decimal = df.copy()

for column in ['Age', 'Salary', 'Balance']:
    max_value = df_decimal[column].abs().max()

    if max_value == 0:
        continue

    j = len(str(int(max_value)))

    df_decimal[column] = df_decimal[column] / (10 ** j)

print("\n4. DECIMAL SCALING")
display(df_decimal[['Age', 'Salary', 'Balance']])


# ------------------------------------------------------------
# 5. ROBUST SCALING
# ------------------------------------------------------------

# Uses median and interquartile range.
# Useful when data contains outliers.

robust = RobustScaler()

df_robust = df.copy()

df_robust[['Age', 'Salary', 'Balance']] = (
    robust.fit_transform(
        df_robust[['Age', 'Salary', 'Balance']]
    )
)

print("\n5. ROBUST SCALING")
display(df_robust[['Age', 'Salary', 'Balance']])


# ------------------------------------------------------------
# 6. LABEL ENCODING
# ------------------------------------------------------------

# Converts categories into numerical labels.

df_label = df.copy()

encoder = LabelEncoder()

df_label['Gender_Encoded'] = encoder.fit_transform(
    df_label['Gender']
)

print("\n6. LABEL ENCODING")
display(df_label[['Gender', 'Gender_Encoded']])


# ------------------------------------------------------------
# 7. ONE-HOT ENCODING
# ------------------------------------------------------------

# Creates separate binary columns for each category.

df_onehot = pd.get_dummies(
    df,
    columns=['City'],
    dtype=int
)

print("\n7. ONE-HOT ENCODING")
display(df_onehot)


# ------------------------------------------------------------
# 8. ORDINAL ENCODING
# ------------------------------------------------------------

# Converts ordered categories into numerical values.
# Education has a meaningful order.

df_ordinal = df.copy()

education_order = [
    'Undergraduate',
    'Graduate',
    'Postgraduate'
]

ordinal = OrdinalEncoder(
    categories=[education_order]
)

df_ordinal['Education_Encoded'] = ordinal.fit_transform(
    df_ordinal[['Education']]
).astype(int)

print("\n8. ORDINAL ENCODING")
display(df_ordinal[['Education', 'Education_Encoded']])


# ------------------------------------------------------------
# 9. BINARY ENCODING
# ------------------------------------------------------------

# Converts category labels into binary representations.
# Uses category codes as an illustrative binary encoding.

df_binary = df.copy()

category_codes = pd.Categorical(
    df_binary['City']
).codes

binary_values = [
    format(value, '02b')
    for value in category_codes
]

df_binary['City_Binary'] = binary_values

print("\n9. BINARY ENCODING")
display(df_binary[['City', 'City_Binary']])


# ------------------------------------------------------------
# 10. FREQUENCY ENCODING
# ------------------------------------------------------------

# Replaces categories with their occurrence frequency.

df_frequency = df.copy()

frequency = df_frequency['City'].value_counts()

df_frequency['City_Frequency'] = (
    df_frequency['City'].map(frequency)
)

print("\n10. FREQUENCY ENCODING")
display(df_frequency[['City', 'City_Frequency']])


# ------------------------------------------------------------
# 11. LOG TRANSFORMATION
# ------------------------------------------------------------

# Reduces right-skewness in positive numerical data.
# log1p handles zero values safely.

df_log = df.copy()

df_log['Salary_Log'] = np.log1p(df_log['Salary'])

df_log['Balance_Log'] = np.log1p(df_log['Balance'])

print("\n11. LOG TRANSFORMATION")
display(df_log[['Salary', 'Salary_Log', 'Balance', 'Balance_Log']])


# ------------------------------------------------------------
# 12. SQUARE ROOT TRANSFORMATION
# ------------------------------------------------------------

# Applies square root to reduce skewness.

df_sqrt = df.copy()

df_sqrt['Salary_Sqrt'] = np.sqrt(df_sqrt['Salary'])

print("\n12. SQUARE ROOT TRANSFORMATION")
display(df_sqrt[['Salary', 'Salary_Sqrt']])


# ------------------------------------------------------------
# 13. POWER TRANSFORMATION
# ------------------------------------------------------------

# Yeo-Johnson transformation can handle positive,
# zero and negative values.

power = PowerTransformer(
    method='yeo-johnson',
    standardize=True
)

df_power = df.copy()

df_power['Salary_Power'] = power.fit_transform(
    df_power[['Salary']]
)

print("\n13. POWER TRANSFORMATION")
display(df_power[['Salary', 'Salary_Power']])


# ------------------------------------------------------------
# 14. FEATURE CONSTRUCTION
# ------------------------------------------------------------

# Creates new features using existing columns.

df_feature = df.copy()

# Annual salary from monthly salary.

df_feature['Annual_Salary'] = (
    df_feature['Salary'] * 12
)

# Total financial exposure.

df_feature['Total_Exposure'] = (
    df_feature['Balance'] +
    df_feature['Loan_Amount']
)

print("\n14. FEATURE CONSTRUCTION")
display(
    df_feature[
        ['Salary', 'Annual_Salary',
         'Balance', 'Loan_Amount', 'Total_Exposure']
    ]
)


# ------------------------------------------------------------
# 15. FEATURE EXTRACTION
# ------------------------------------------------------------

# Extracts useful information from existing columns.

df_extraction = df.copy()

# Extract first letter from customer name.

df_extraction['Name_Initial'] = (
    df_extraction['Name'].str[0]
)

# Extract first three letters from city.

df_extraction['City_Code'] = (
    df_extraction['City'].str[:3].str.upper()
)

print("\n15. FEATURE EXTRACTION")
display(
    df_extraction[
        ['Name', 'Name_Initial', 'City', 'City_Code']
    ]
)


# ------------------------------------------------------------
# 16. DATA TYPE CONVERSION
# ------------------------------------------------------------

# Converts columns into suitable data types.

df_types = df.copy()

df_types['Customer_ID'] = (
    df_types['Customer_ID'].astype(str)
)

df_types['Age'] = (
    df_types['Age'].astype(float)
)

print("\n16. DATA TYPE CONVERSION")
display(df_types.dtypes)


# ------------------------------------------------------------
# 17. DATE AND TIME TRANSFORMATION
# ------------------------------------------------------------

# Converts text dates into datetime format.
# Extracts year, month and day.

df_date = df.copy()

df_date['Joining_Date'] = pd.to_datetime(
    df_date['Joining_Date']
)

df_date['Joining_Year'] = (
    df_date['Joining_Date'].dt.year
)

df_date['Joining_Month'] = (
    df_date['Joining_Date'].dt.month
)

df_date['Joining_Day'] = (
    df_date['Joining_Date'].dt.day
)

print("\n17. DATE AND TIME TRANSFORMATION")
display(
    df_date[
        ['Joining_Date', 'Joining_Year',
         'Joining_Month', 'Joining_Day']
    ]
)


# ------------------------------------------------------------
# 18. AGGREGATION TRANSFORMATION
# ------------------------------------------------------------

# Summarizes customer salary and balance by city.

df_aggregation = df.copy()

aggregated_data = df_aggregation.groupby('City').agg({
    'Salary': 'mean',
    'Balance': 'sum',
    'Loan_Amount': 'sum'
}).reset_index()

print("\n18. AGGREGATION TRANSFORMATION")
display(aggregated_data)


# ------------------------------------------------------------
# 19. BINNING TRANSFORMATION
# ------------------------------------------------------------

# Converts continuous age values into categories.

df_binning = df.copy()

df_binning['Age_Group'] = pd.cut(
    df_binning['Age'],
    bins=[0, 25, 40, 60],
    labels=['Young', 'Adult', 'Senior']
)

print("\n19. BINNING TRANSFORMATION")
display(df_binning[['Age', 'Age_Group']])


# ------------------------------------------------------------
# 20. STRING TRANSFORMATION
# ------------------------------------------------------------

# Standardizes text values.

df_string = df.copy()

df_string['Name'] = (
    df_string['Name'].str.upper()
)

df_string['City'] = (
    df_string['City'].str.lower()
)

df_string['Gender'] = (
    df_string['Gender'].str.strip()
)

print("\n20. STRING TRANSFORMATION")
display(
    df_string[['Name', 'City', 'Gender']]
)


# ------------------------------------------------------------
# 21. FUNCTION-BASED TRANSFORMATION
# ------------------------------------------------------------

# Applies a custom function to transform values.

df_function = df.copy()

def salary_category(salary):

    if salary < 30000:
        return 'Low'

    elif salary < 60000:
        return 'Medium'

    else:
        return 'High'


df_function['Salary_Category'] = (
    df_function['Salary'].apply(salary_category)
)

print("\n21. FUNCTION-BASED TRANSFORMATION")
display(
    df_function[['Salary', 'Salary_Category']]
)


# ------------------------------------------------------------
# 22. FINAL DATA TRANSFORMATION SUMMARY
# ------------------------------------------------------------

print("\n========================================")
print("     DATA TRANSFORMATION SUMMARY")
print("========================================")

print("Original Dataset Shape:", df.shape)

print("Min-Max Normalization:", df_minmax.shape)

print("Standardization:", df_standard.shape)

print("Decimal Scaling:", df_decimal.shape)

print("Robust Scaling:", df_robust.shape)

print("Label Encoding:", df_label.shape)

print("One-Hot Encoding:", df_onehot.shape)

print("Ordinal Encoding:", df_ordinal.shape)

print("Binary Encoding:", df_binary.shape)

print("Frequency Encoding:", df_frequency.shape)

print("Log Transformation:", df_log.shape)

print("Square Root Transformation:", df_sqrt.shape)

print("Power Transformation:", df_power.shape)

print("Feature Construction:", df_feature.shape)

print("Feature Extraction:", df_extraction.shape)

print("Date Transformation:", df_date.shape)

print("Aggregation:", aggregated_data.shape)

print("Binning:", df_binning.shape)

print("\nALL DATA TRANSFORMATION TECHNIQUES COMPLETED!")
```
