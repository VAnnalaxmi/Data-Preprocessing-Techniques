```python

# ============================================================
# DATA INTEGRATION - ALL MAJOR TECHNIQUES
# Banking Customer Dataset
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. CREATE BANKING DATASETS
# ------------------------------------------------------------

# Customer information

customers = pd.DataFrame({
    'Customer_ID': [101, 102, 103, 104, 105],
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'John'],
    'Age': [22, 35, 28, 45, 32],
    'City': ['Hyderabad', 'Chennai', 'Delhi', 'Mumbai', 'Hyderabad']
})

# Account information

accounts = pd.DataFrame({
    'Customer_ID': [101, 102, 103, 104, 106],
    'Account_Type': ['Savings', 'Current', 'Savings', 'Current', 'Savings'],
    'Balance': [10000, 25000, 15000, 50000, 30000]
})

# Loan information

loans = pd.DataFrame({
    'Customer_ID': [101, 103, 104, 105, 107],
    'Loan_Amount': [50000, 20000, 100000, 30000, 60000]
})

print("CUSTOMER DATASET")
display(customers)

print("ACCOUNT DATASET")
display(accounts)

print("LOAN DATASET")
display(loans)


# ------------------------------------------------------------
# 2. DATA MERGING
# ------------------------------------------------------------

# Combines two datasets using a common Customer_ID.

merged_data = pd.merge(
    customers,
    accounts,
    on='Customer_ID'
)

print("\n2. DATA MERGING")
display(merged_data)


# ------------------------------------------------------------
# 3. INNER JOIN
# ------------------------------------------------------------

# Returns only records having matching Customer_ID
# in both datasets.

inner_join = pd.merge(
    customers,
    accounts,
    on='Customer_ID',
    how='inner'
)

print("\n3. INNER JOIN")
display(inner_join)


# ------------------------------------------------------------
# 4. LEFT JOIN
# ------------------------------------------------------------

# Keeps all records from the left dataset (customers).
# Non-matching account details become NaN.

left_join = pd.merge(
    customers,
    accounts,
    on='Customer_ID',
    how='left'
)

print("\n4. LEFT JOIN")
display(left_join)


# ------------------------------------------------------------
# 5. RIGHT JOIN
# ------------------------------------------------------------

# Keeps all records from the right dataset (accounts).

right_join = pd.merge(
    customers,
    accounts,
    on='Customer_ID',
    how='right'
)

print("\n5. RIGHT JOIN")
display(right_join)


# ------------------------------------------------------------
# 6. FULL OUTER JOIN
# ------------------------------------------------------------

# Keeps matching and non-matching records
# from both datasets.

outer_join = pd.merge(
    customers,
    accounts,
    on='Customer_ID',
    how='outer'
)

print("\n6. FULL OUTER JOIN")
display(outer_join)


# ------------------------------------------------------------
# 7. CROSS JOIN
# ------------------------------------------------------------

# Combines every customer record with every account type.
# This produces all possible combinations.

account_types = pd.DataFrame({
    'Account_Type': ['Savings', 'Current']
})

cross_join = pd.merge(
    customers,
    account_types,
    how='cross'
)

print("\n7. CROSS JOIN")
display(cross_join)


# ------------------------------------------------------------
# 8. ROW-WISE CONCATENATION
# ------------------------------------------------------------

# Combines customer records from different branches.

branch_1 = pd.DataFrame({
    'Customer_ID': [101, 102],
    'Name': ['Anu', 'Ravi']
})

branch_2 = pd.DataFrame({
    'Customer_ID': [103, 104],
    'Name': ['Priya', 'Kiran']
})

row_concat = pd.concat(
    [branch_1, branch_2],
    axis=0,
    ignore_index=True
)

print("\n8. ROW-WISE CONCATENATION")
display(row_concat)


# ------------------------------------------------------------
# 9. COLUMN-WISE CONCATENATION
# ------------------------------------------------------------

# Combines columns from two datasets based on row index.
# Both datasets must have matching row order.

customer_names = customers[['Customer_ID', 'Name']]

customer_ages = customers[['Age', 'City']]

column_concat = pd.concat(
    [customer_names, customer_ages],
    axis=1
)

print("\n9. COLUMN-WISE CONCATENATION")
display(column_concat)


# ------------------------------------------------------------
# 10. INDEX-BASED JOINING
# ------------------------------------------------------------

# Sets Customer_ID as the index and joins both datasets.

customer_index = customers.set_index('Customer_ID')

account_index = accounts.set_index('Customer_ID')

index_join = customer_index.join(
    account_index,
    how='left'
)

print("\n10. INDEX-BASED JOINING")
display(index_join)


# ------------------------------------------------------------
# 11. MULTIPLE DATASET INTEGRATION
# ------------------------------------------------------------

# First combines customers and accounts.
# Then combines the result with loan information.

customer_account = pd.merge(
    customers,
    accounts,
    on='Customer_ID',
    how='left'
)

complete_data = pd.merge(
    customer_account,
    loans,
    on='Customer_ID',
    how='left'
)

print("\n11. MULTIPLE DATASET INTEGRATION")
display(complete_data)


# ------------------------------------------------------------
# 12. SCHEMA INTEGRATION
# ------------------------------------------------------------

# Different datasets may use different column names.
# Rename them into a common schema before combining.

branch_a = pd.DataFrame({
    'Cust_ID': [201, 202],
    'Cust_Name': ['Meena', 'Arun'],
    'Location': ['Hyderabad', 'Chennai']
})

branch_b = pd.DataFrame({
    'Customer_ID': [203, 204],
    'Name': ['Sita', 'Rahul'],
    'City': ['Mumbai', 'Delhi']
})

# Standardize column names.

branch_a = branch_a.rename(columns={
    'Cust_ID': 'Customer_ID',
    'Cust_Name': 'Name',
    'Location': 'City'
})

branch_b = branch_b.rename(columns={
    'Customer_ID': 'Customer_ID',
    'Name': 'Name',
    'City': 'City'
})

# Combine both datasets.

schema_integrated = pd.concat(
    [branch_a, branch_b],
    ignore_index=True
)

print("\n12. SCHEMA INTEGRATION")
display(schema_integrated)


# ------------------------------------------------------------
# 13. DATA TYPE INTEGRATION
# ------------------------------------------------------------

# Converts columns into consistent data types
# before integrating different sources.

source_a = pd.DataFrame({
    'Customer_ID': ['301', '302'],
    'Salary': ['25000', '35000']
})

source_b = pd.DataFrame({
    'Customer_ID': [303, 304],
    'Salary': [45000, 55000]
})

# Convert both datasets to the same data types.

source_a['Customer_ID'] = source_a['Customer_ID'].astype(int)
source_a['Salary'] = source_a['Salary'].astype(float)

source_b['Customer_ID'] = source_b['Customer_ID'].astype(int)
source_b['Salary'] = source_b['Salary'].astype(float)

type_integrated = pd.concat(
    [source_a, source_b],
    ignore_index=True
)

print("\n13. DATA TYPE INTEGRATION")
display(type_integrated)

print(type_integrated.dtypes)


# ------------------------------------------------------------
# 14. ENTITY RESOLUTION
# ------------------------------------------------------------

# Identifies records referring to the same customer
# even when names have different capitalization or spaces.

source_customers = pd.DataFrame({
    'Customer_ID': [401, 402, 403],
    'Name': [' ANU ', 'RAVI', 'Priya']
})

# Standardize names before matching.

source_customers['Standard_Name'] = (
    source_customers['Name']
    .str.strip()
    .str.lower()
)

print("\n14. ENTITY RESOLUTION")
display(source_customers)


# ------------------------------------------------------------
# 15. DUPLICATE RECORD IDENTIFICATION
# ------------------------------------------------------------

# Identifies repeated customer records.

duplicate_customers = pd.DataFrame({
    'Customer_ID': [501, 502, 501, 503],
    'Name': ['Anu', 'Ravi', 'Anu', 'Priya']
})

print("\n15. DUPLICATE RECORD IDENTIFICATION")

print("Duplicate Records:")
display(
    duplicate_customers[
        duplicate_customers.duplicated(
            subset=['Customer_ID'],
            keep=False
        )
    ]
)

# Removes duplicate customer IDs.

unique_customers = duplicate_customers.drop_duplicates(
    subset=['Customer_ID'],
    keep='first'
)

print("Unique Customer Records:")
display(unique_customers)


# ------------------------------------------------------------
# 16. CONFLICT RESOLUTION
# ------------------------------------------------------------

# Different sources may contain different values
# for the same customer.

source_1 = pd.DataFrame({
    'Customer_ID': [601, 602],
    'City': ['Hyderabad', 'Chennai']
})

source_2 = pd.DataFrame({
    'Customer_ID': [601, 602],
    'City': ['Hyderabad', 'Mumbai']
})

# Merge both sources to compare conflicting values.

conflict_data = pd.merge(
    source_1,
    source_2,
    on='Customer_ID',
    suffixes=('_Source1', '_Source2')
)

# Detect conflicting city values.

conflict_data['Conflict'] = (
    conflict_data['City_Source1'] !=
    conflict_data['City_Source2']
)

print("\n16. CONFLICT RESOLUTION")
display(conflict_data)

# Example resolution: prefer Source 1.
# In real systems, the preferred source should be
# decided using data governance rules.

conflict_data['Final_City'] = conflict_data['City_Source1']

display(
    conflict_data[
        ['Customer_ID', 'Final_City', 'Conflict']
    ]
)


# ------------------------------------------------------------
# 17. DATA AGGREGATION
# ------------------------------------------------------------

# Combines detailed records into summarized information.

branch_data = pd.DataFrame({
    'City': ['Hyderabad', 'Hyderabad', 'Chennai',
             'Chennai', 'Mumbai'],
    'Balance': [10000, 20000, 15000, 25000, 30000]
})

aggregated_data = branch_data.groupby('City').agg({
    'Balance': ['sum', 'mean', 'count']
})

print("\n17. DATA AGGREGATION")
display(aggregated_data)


# ------------------------------------------------------------
# 18. REFERENTIAL INTEGRITY CHECKING
# ------------------------------------------------------------

# Checks whether account Customer_ID values
# exist in the customer master dataset.

invalid_accounts = accounts[
    ~accounts['Customer_ID'].isin(
        customers['Customer_ID']
    )
]

print("\n18. REFERENTIAL INTEGRITY CHECKING")
print("Accounts Without Matching Customers:")
display(invalid_accounts)


# ------------------------------------------------------------
# 19. FINAL INTEGRATED DATASET
# ------------------------------------------------------------

# Combines customer, account and loan datasets.
# One row per customer is maintained because
# each source has one record per Customer_ID.

final_data = pd.merge(
    customers,
    accounts,
    on='Customer_ID',
    how='left',
    validate='one_to_one'
)

final_data = pd.merge(
    final_data,
    loans,
    on='Customer_ID',
    how='left',
    validate='one_to_one'
)

print("\n19. FINAL INTEGRATED DATASET")
display(final_data)


# ------------------------------------------------------------
# 20. FINAL INTEGRATION SUMMARY
# ------------------------------------------------------------

print("\n========================================")
print("       DATA INTEGRATION SUMMARY")
print("========================================")

print("Customer Dataset Shape:", customers.shape)

print("Account Dataset Shape:", accounts.shape)

print("Loan Dataset Shape:", loans.shape)

print("Inner Join Shape:", inner_join.shape)

print("Left Join Shape:", left_join.shape)

print("Right Join Shape:", right_join.shape)

print("Outer Join Shape:", outer_join.shape)

print("Row Concatenation Shape:", row_concat.shape)

print("Column Concatenation Shape:", column_concat.shape)

print("Multiple Dataset Integration Shape:",
      complete_data.shape)

print("Final Integrated Dataset Shape:",
      final_data.shape)

print("\nALL DATA INTEGRATION TECHNIQUES COMPLETED!")
```
