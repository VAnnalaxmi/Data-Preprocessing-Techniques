```python

#data cleaning
# ==========================================================
# DATA CLEANING: MISSING VALUE HANDLING
# Dataset: Banking Customer Details
# ==========================================================

import pandas as pd
import numpy as np

# STEP 1: Create a sample banking dataset
# NaN represents missing values.

data = {
    'Customer_ID': [101, 102, 103, 104, 105, 106],
    'Age': [25, 30, np.nan, 40, 35, np.nan],
    'Salary': [30000, np.nan, 50000, 60000, np.nan, 40000],
    'Gender': ['Male', 'Female', np.nan, 'Male', 'Female', np.nan],
    'City': ['Hyderabad', 'Chennai', 'Mumbai', np.nan, 'Delhi', 'Chennai']
}

df = pd.DataFrame(data)

print("ORIGINAL DATASET")
print(df)


# STEP 2: Identify missing values
# isnull() identifies missing values.
# sum() counts missing values in each column.

print("\nMISSING VALUES")
print(df.isnull().sum())


# ==========================================================
# TECHNIQUE 1: IGNORING / DELETING MISSING VALUES
# Remove rows containing missing values using dropna().
# ==========================================================

df_ignore = df.dropna()

print("\n1. AFTER DELETING MISSING VALUES")
print(df_ignore)


# ==========================================================
# TECHNIQUE 2: CONSTANT VALUE IMPUTATION
# Replace missing values with a fixed value.
# Numerical columns: 0
# Categorical columns: Unknown
# ==========================================================

df_constant = df.copy()

df_constant['Age'] = df_constant['Age'].fillna(0)
df_constant['Salary'] = df_constant['Salary'].fillna(0)
df_constant['Gender'] = df_constant['Gender'].fillna('Unknown')
df_constant['City'] = df_constant['City'].fillna('Unknown')

print("\n2. AFTER CONSTANT VALUE IMPUTATION")
print(df_constant)


# ==========================================================
# TECHNIQUE 3: MEAN IMPUTATION
# Replace missing numerical values with their average.
# ==========================================================

df_mean = df.copy()

df_mean['Age'] = df_mean['Age'].fillna(df_mean['Age'].mean())
df_mean['Salary'] = df_mean['Salary'].fillna(df_mean['Salary'].mean())

print("\n3. AFTER MEAN IMPUTATION")
print(df_mean)


# ==========================================================
# TECHNIQUE 4: MEDIAN IMPUTATION
# Replace missing numerical values with their middle value.
# ==========================================================

df_median = df.copy()

df_median['Age'] = df_median['Age'].fillna(df_median['Age'].median())
df_median['Salary'] = df_median['Salary'].fillna(df_median['Salary'].median())

print("\n4. AFTER MEDIAN IMPUTATION")
print(df_median)


# ==========================================================
# TECHNIQUE 5: MODE IMPUTATION
# Replace missing values with the most frequently occurring value.
# Suitable for categorical columns such as Gender and City.
# ==========================================================

df_mode = df.copy()

df_mode['Gender'] = df_mode['Gender'].fillna(
    df_mode['Gender'].mode()[0]
)

df_mode['City'] = df_mode['City'].fillna(
    df_mode['City'].mode()[0]
)

print("\n5. AFTER MODE IMPUTATION")
print(df_mode)


# ==========================================================
# FINAL COMPARISON
# Check missing values after applying each technique.
# ==========================================================

print("\nMISSING VALUES AFTER EACH TECHNIQUE")

print("Constant:\n", df_constant.isnull().sum())

print("Mean:\n", df_mean.isnull().sum())

print("Median:\n", df_median.isnull().sum())

print("Mode:\n", df_mode.isnull().sum())

print("\nMISSING VALUE HANDLING COMPLETED!")


```
